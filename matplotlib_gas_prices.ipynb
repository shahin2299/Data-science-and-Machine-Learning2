# Matplotlib দিয়ে Gas Prices Dataset Visualization

এই নোটবুকে প্রশ্ন অনুযায়ী matplotlib দিয়ে বিভিন্ন ধরনের chart/graph বানানো শেখানো হয়েছে। প্রতিটি cell আলাদাভাবে top to bottom রান করলেই কাজ করবে।

**নোট:** প্রথমে Colab এ `gas_prices.csv` ফাইলটি upload করে নিন।

```python
# প্রয়োজনীয় লাইব্রেরি import ও dataset load
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('gas_prices.csv')
df.head()
```

## প্রশ্ন ১: Line Plot দিয়ে USA এর gas price এর trend (বছর অনুযায়ী) দেখাও

```python
# plot() দিয়ে line chart বানানো হলো
plt.plot(df['Year'], df['USA'])
plt.xlabel('Year')
plt.ylabel('Gas Price (USA)')
plt.title('USA Gas Price Over Years')
plt.show()
```

## প্রশ্ন ২: একই plot এ USA, Canada, UK — তিনটি দেশের line plot compare করো

```python
# একই figure এ একাধিক line আঁকা হলো এবং label দিয়ে legend বানানো হলো
plt.plot(df['Year'], df['USA'], label='USA')
plt.plot(df['Year'], df['Canada'], label='Canada')
plt.plot(df['Year'], df['UK'], label='UK')
plt.xlabel('Year')
plt.ylabel('Gas Price')
plt.title('Gas Price Comparison')
plt.legend()
plt.show()
```

## প্রশ্ন ৩: Bar Chart দিয়ে প্রতিটি দেশের গড় (average) gas price দেখাও

```python
# প্রথমে প্রতি দেশের গড় বের করে bar() দিয়ে চার্ট বানানো হলো
avg_prices = df.drop(columns='Year').mean(numeric_only=True)
plt.bar(avg_prices.index, avg_prices.values)
plt.xticks(rotation=45)
plt.ylabel('Average Gas Price')
plt.title('Average Gas Price by Country')
plt.show()
```

## প্রশ্ন ৪: Horizontal Bar Chart দিয়ে একই তথ্য দেখাও

```python
# barh() দিয়ে horizontal bar chart বানানো হলো
avg_prices = df.drop(columns='Year').mean(numeric_only=True)
plt.barh(avg_prices.index, avg_prices.values)
plt.xlabel('Average Gas Price')
plt.title('Average Gas Price by Country (Horizontal)')
plt.show()
```

## প্রশ্ন ৫: Histogram দিয়ে USA এর gas price এর distribution দেখাও

```python
# hist() দিয়ে value গুলোর frequency distribution দেখানো হলো
plt.hist(df['USA'], bins=8, color='skyblue', edgecolor='black')
plt.xlabel('Gas Price (USA)')
plt.ylabel('Frequency')
plt.title('Distribution of USA Gas Price')
plt.show()
```

## প্রশ্ন ৬: Scatter Plot দিয়ে USA আর Canada এর price এর মধ্যে সম্পর্ক দেখাও

```python
# scatter() দিয়ে দুইটা variable এর মধ্যে সম্পর্ক দেখানো হলো
plt.scatter(df['USA'], df['Canada'])
plt.xlabel('USA')
plt.ylabel('Canada')
plt.title('USA vs Canada Gas Price')
plt.show()
```

## প্রশ্ন ৭: Pie Chart দিয়ে সর্বশেষ বছরে দেশগুলোর gas price এর share দেখাও

```python
# dataset এর শেষ row (সর্বশেষ বছর) নিয়ে pie chart বানানো হলো
last_year = df.iloc[-1]
values = last_year.drop('Year')
plt.pie(values, labels=values.index, autopct='%1.1f%%')
plt.title(f"Gas Price Share in {int(last_year['Year'])}")
plt.show()
```

## প্রশ্ন ৮: Subplot ব্যবহার করে একসাথে দুইটা দেশের আলাদা আলাদা line plot দেখাও

```python
# subplots() দিয়ে একই figure এ একাধিক আলাদা plot বানানো হলো
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].plot(df['Year'], df['USA'], color='blue')
axes[0].set_title('USA')
axes[1].plot(df['Year'], df['Japan'], color='green')
axes[1].set_title('Japan')
plt.tight_layout()
plt.show()
```

## প্রশ্ন ৯: Box Plot দিয়ে সব দেশের gas price এর distribution compare করো

```python
# boxplot() দিয়ে outlier ও spread একসাথে দেখা যায়
data = df.drop(columns='Year')
plt.boxplot(data.dropna().values, tick_labels=data.columns)
plt.xticks(rotation=45)
plt.ylabel('Gas Price')
plt.title('Gas Price Distribution by Country')
plt.show()
```

## প্রশ্ন ১০: Plot কাস্টমাইজ করো (color, linestyle, marker, grid ব্যবহার করে)

```python
# color, linestyle, marker, grid দিয়ে plot কে customize করা হলো
plt.plot(df['Year'], df['Germany'], color='red', linestyle='--', marker='o')
plt.grid(True)
plt.title('Germany Gas Price (Customized)')
plt.xlabel('Year')
plt.ylabel('Price')
plt.show()
```

## প্রশ্ন ১১: figure টি একটি image ফাইল হিসেবে save করো

```python
# savefig() দিয়ে plot কে png ফাইলে save করা হলো
plt.plot(df['Year'], df['USA'])
plt.title('USA Gas Price')
plt.savefig('usa_gas_price.png')
plt.show()
```
