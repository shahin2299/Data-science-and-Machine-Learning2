# Seaborn দিয়ে Gas Prices Dataset Visualization

এই নোটবুকে প্রশ্ন অনুযায়ী seaborn দিয়ে বিভিন্ন ধরনের statistical visualization বানানো শেখানো হয়েছে। প্রতিটি cell আলাদাভাবে top to bottom রান করলেই কাজ করবে।

**নোট:** প্রথমে Colab এ `gas_prices.csv` ফাইলটি upload করে নিন।

```python
# প্রয়োজনীয় লাইব্রেরি import ও dataset load
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('gas_prices.csv')
df.head()
```

## প্রশ্ন ১: seaborn এর lineplot দিয়ে USA এর gas price এর trend দেখাও

```python
# sns.lineplot() দিয়ে trend দেখানো হলো
sns.lineplot(x='Year', y='USA', data=df)
plt.title('USA Gas Price Trend (Seaborn)')
plt.show()
```

## প্রশ্ন ২: dataset কে wide থেকে long format এ রূপান্তর করো (melt)

```python
# একাধিক দেশ একসাথে plot করার জন্য melt() দিয়ে long format বানানো হলো
df_long = df.melt(id_vars='Year', var_name='Country', value_name='Price')
df_long.head()
```

## প্রশ্ন ৩: seaborn lineplot এ hue ব্যবহার করে সব দেশের trend একসাথে compare করো

```python
# hue='Country' দিলে প্রতিটি দেশ আলাদা রঙের line হিসেবে দেখা যায়
plt.figure(figsize=(10, 6))
sns.lineplot(x='Year', y='Price', hue='Country', data=df_long)
plt.title('Gas Price Trend - All Countries')
plt.show()
```

## প্রশ্ন ৪: seaborn barplot দিয়ে প্রতিটি দেশের গড় gas price দেখাও

```python
# sns.barplot() default ভাবে mean দেখায় এবং confidence interval ও দেখায়
plt.figure(figsize=(10, 5))
sns.barplot(x='Country', y='Price', data=df_long)
plt.xticks(rotation=45)
plt.title('Average Gas Price by Country')
plt.show()
```

## প্রশ্ন ৫: seaborn boxplot দিয়ে দেশভেদে price এর distribution দেখাও

```python
# boxplot দিয়ে median, quartile, outlier দেখা যায়
plt.figure(figsize=(10, 5))
sns.boxplot(x='Country', y='Price', data=df_long)
plt.xticks(rotation=45)
plt.title('Gas Price Distribution by Country (Boxplot)')
plt.show()
```

## প্রশ্ন ৬: seaborn violinplot দিয়ে একই তথ্য দেখাও (distribution shape সহ)

```python
# violinplot boxplot এর মতোই কিন্তু distribution এর shape ও দেখায়
plt.figure(figsize=(10, 5))
sns.violinplot(x='Country', y='Price', data=df_long)
plt.xticks(rotation=45)
plt.title('Gas Price Distribution by Country (Violinplot)')
plt.show()
```

## প্রশ্ন ৭: seaborn histplot দিয়ে USA এর price এর distribution দেখাও (density curve সহ)

```python
# kde=True দিলে histogram এর সাথে density curve ও দেখা যায়
sns.histplot(df['USA'], kde=True)
plt.title('USA Gas Price Distribution')
plt.show()
```

## প্রশ্ন ৮: seaborn heatmap দিয়ে দেশগুলোর মধ্যে correlation দেখাও

```python
# heatmap দিয়ে correlation matrix visually দেখানো হলো
plt.figure(figsize=(8, 6))
sns.heatmap(df.drop(columns='Year').corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

## প্রশ্ন ৯: seaborn regplot দিয়ে USA আর Canada এর মধ্যে সম্পর্ক ও regression line দেখাও

```python
# regplot scatter plot এর সাথে best-fit regression line ও দেখায়
sns.regplot(x='USA', y='Canada', data=df)
plt.title('USA vs Canada (with Regression Line)')
plt.show()
```

## প্রশ্ন ১০: seaborn pairplot দিয়ে কয়েকটি দেশের মধ্যে সম্পর্ক একসাথে দেখাও

```python
# pairplot দিয়ে একাধিক column এর মধ্যে pairwise সম্পর্ক একসাথে দেখা যায়
sns.pairplot(df[['USA', 'Canada', 'UK', 'Japan']])
plt.show()
```

## প্রশ্ন ১১: seaborn theme/style ব্যবহার করে plot এর appearance পরিবর্তন করো

```python
# set_style() দিয়ে plot এর background style পরিবর্তন করা হলো
sns.set_style('darkgrid')
sns.lineplot(x='Year', y='USA', data=df)
plt.title('Styled Plot')
plt.show()
```
