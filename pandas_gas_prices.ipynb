# Pandas দিয়ে Gas Prices Dataset Analysis

এই নোটবুকে প্রশ্ন অনুযায়ী ধাপে ধাপে pandas এর গুরুত্বপূর্ণ ফাংশনগুলো শেখানো হয়েছে। প্রতিটি প্রশ্নের নিচে উত্তরের code cell আছে — প্রতিটি cell আলাদাভাবে (top to bottom) রান করলেই কাজ করবে।

**নোট:** প্রথমে Colab এ বাম পাশের Files প্যানেল থেকে `gas_prices.csv` ফাইলটি upload করে নিন।

```python
# pandas লাইব্রেরি import করা হলো
import pandas as pd
```

## প্রশ্ন ১: CSV ফাইল থেকে dataset load করে দেখাও

```python
# CSV ফাইল থেকে data load করে DataFrame বানানো হলো
df = pd.read_csv('gas_prices.csv')
df
```

## প্রশ্ন ২: dataset এর প্রথম ৫টি row দেখাও (head)

```python
# head() দিয়ে শুরুর ৫টি row দেখা যায়
df.head()
```

## প্রশ্ন ৩: dataset এর শেষ ৫টি row দেখাও (tail)

```python
# tail() দিয়ে শেষের ৫টি row দেখা যায়
df.tail()
```

## প্রশ্ন ৪: dataset এ মোট কতগুলো row ও column আছে তা বের করো

```python
# shape দিয়ে (row সংখ্যা, column সংখ্যা) পাওয়া যায়
df.shape
```

## প্রশ্ন ৫: dataset এর সব column এর নাম দেখাও

```python
# columns attribute দিয়ে সব column এর নাম দেখা যায়
df.columns
```

## প্রশ্ন ৬: প্রতিটি column এর data type ও non-null value সংখ্যা দেখাও

```python
# info() দিয়ে dtype, non-null count ইত্যাদি দেখা যায়
df.info()
```

## প্রশ্ন ৭: dataset এর statistical summary (mean, std, min, max ইত্যাদি) দেখাও

```python
# describe() দিয়ে numeric column গুলোর statistics দেখা যায়
df.describe()
```

## প্রশ্ন ৮: কোন কোন column এ missing value (null) আছে ও কয়টা আছে তা বের করো

```python
# isnull().sum() দিয়ে প্রতি column এ missing value সংখ্যা গণনা করা হলো
df.isnull().sum()
```

## প্রশ্ন ৯: missing value গুলো সেই column এর গড় (mean) দিয়ে পূরণ করো

```python
# fillna() দিয়ে missing value গুলো column এর mean দিয়ে পূরণ করা হলো
df = df.fillna(df.mean(numeric_only=True))
df.isnull().sum()
```

## প্রশ্ন ১০: শুধু 'USA' column টি select করে দেখাও

```python
# একটি column select করার নিয়ম
df['USA']
```

## প্রশ্ন ১১: 'Year', 'USA', 'Canada' — এই তিনটি column একসাথে select করো

```python
# একাধিক column select করতে list of column name দিতে হয়
df[['Year', 'USA', 'Canada']]
```

## প্রশ্ন ১২: 2000 সালের পরের (Year > 2000) সব row filter করে দেখাও

```python
# condition দিয়ে row filter করা
df[df['Year'] > 2000]
```

## প্রশ্ন ১৩: যেসব বছরে USA তে gas price 1.5 এর বেশি ছিল সেগুলো filter করো

```python
# একাধিক condition ছাড়াও single condition দিয়ে filter করা যায়
df[df['USA'] > 1.5]
```

## প্রশ্ন ১৪: USA column অনুযায়ী dataset কে ছোট থেকে বড় (ascending) sort করো

```python
# sort_values() দিয়ে column অনুযায়ী sort করা যায়
df.sort_values(by='USA')
```

## প্রশ্ন ১৫: USA column অনুযায়ী বড় থেকে ছোট (descending) sort করো

```python
# ascending=False দিলে descending order এ sort হয়
df.sort_values(by='USA', ascending=False)
```

## প্রশ্ন ১৬: নতুন একটি column বানাও যেখানে USA আর Canada এর price এর পার্থক্য থাকবে

```python
# দুইটা column বিয়োগ করে নতুন column তৈরি করা হলো
df['USA_Canada_Diff'] = df['USA'] - df['Canada']
df[['Year', 'USA', 'Canada', 'USA_Canada_Diff']]
```

## প্রশ্ন ১৭: কোন বছরে USA তে সর্বোচ্চ gas price ছিল তা বের করো

```python
# idxmax() দিয়ে সর্বোচ্চ value এর index বের করে সেই row দেখানো হলো
df.loc[df['USA'].idxmax()]
```

## প্রশ্ন ১৮: কোন বছরে USA তে সর্বনিম্ন gas price ছিল তা বের করো

```python
# idxmin() দিয়ে সর্বনিম্ন value এর index বের করা হলো
df.loc[df['USA'].idxmin()]
```

## প্রশ্ন ১৯: 'Year' column কে dataset এর index হিসেবে সেট করো

```python
# set_index() দিয়ে একটি column কে index বানানো যায়
df_indexed = df.set_index('Year')
df_indexed.head()
```

## প্রশ্ন ২০: loc ব্যবহার করে 2005 সালের data দেখাও

```python
# loc[] দিয়ে label (index value) অনুযায়ী row access করা হলো
df_indexed.loc[2005]
```

## প্রশ্ন ২১: iloc ব্যবহার করে প্রথম ৩টি row এবং প্রথম ৪টি column দেখাও

```python
# iloc[] দিয়ে position (index number) অনুযায়ী data access করা হলো
df.iloc[0:3, 0:4]
```

## প্রশ্ন ২২: প্রতিটি country এর গড় (mean) gas price বের করো

```python
# Year বাদ দিয়ে বাকি সব country column এর mean বের করা হলো
df.drop(columns='Year').mean(numeric_only=True)
```

## প্রশ্ন ২৩: সব দেশের মধ্যে correlation (সম্পর্ক) matrix বের করো

```python
# corr() দিয়ে column গুলোর মধ্যে correlation বের করা হলো
df.drop(columns='Year').corr()
```

## প্রশ্ন ২৪: apply() ফাংশন ব্যবহার করে USA এর price কে round করে নতুন column বানাও

```python
# apply() দিয়ে প্রতিটি value তে একটি function চালানো হলো
df['USA_Rounded'] = df['USA'].apply(round)
df[['Year', 'USA', 'USA_Rounded']]
```

## প্রশ্ন ২৫: অপ্রয়োজনীয় column গুলো (যেগুলো আমরা নতুন বানিয়েছিলাম) drop করো

```python
# drop(columns=...) দিয়ে column মুছে ফেলা হলো
df = df.drop(columns=['USA_Canada_Diff', 'USA_Rounded'])
df.head()
```

## প্রশ্ন ২৬: column এর নাম rename করো (যেমন 'USA' কে 'United_States' করো)

```python
# rename() দিয়ে column এর নাম পরিবর্তন করা হলো
df_renamed = df.rename(columns={'USA': 'United_States'})
df_renamed.head()
```

## প্রশ্ন ২৭: groupby ব্যবহার করে প্রতি দশকে (decade) USA এর গড় price বের করো

```python
# Year থেকে Decade বের করে groupby দিয়ে গড় বের করা হলো
df['Decade'] = (df['Year'] // 10) * 10
df.groupby('Decade')['USA'].mean()
```

## প্রশ্ন ২৮: processed dataset টি নতুন CSV ফাইলে save করো

```python
# to_csv() দিয়ে DataFrame কে CSV ফাইলে save করা হলো
df.to_csv('gas_prices_cleaned.csv', index=False)
print('File saved successfully!')
```
